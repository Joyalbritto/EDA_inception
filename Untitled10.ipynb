```python
import spacy

# 1. Load the pre-trained English model
nlp = spacy.load("en_core_web_sm")

# 2. Define the text you want to analyze
text = "APPLE  is looking at buying a U.K. STATUS for $1 billion."

# 3. Process the text to create a 'Doc' object
doc = nlp(text)

# --- Explanation Section ---

print("Tokens and Parts of Speech:")
# Iterate through each word (token) in the processed document
for token in doc:
    # token.text gives the word, token.pos_ gives its part of speech (Noun, Verb, etc.)
    print(f"{token.text} -> {token.pos_}")

print("\nNamed Entities:")
# Iterate through recognized entities (like companies, locations, money)
for ent in doc.ents:
    # ent.text gives the entity, ent.label_ gives its category (ORG, GPE, MONEY, etc.)
    print(f"{ent.text} -> {ent.label_}")
```

```python
import spacy

# 1. Load the pre-trained English model
nlp = spacy.load("en_core_web_sm")

# 2. Define your text
text = "The quick brown fox jumps over the lazy dog in the forest."

# 3. Process the text to create a 'Doc' object
doc = nlp(text)

# 4. Filter out the stop words and punctuation
# We keep the word only if it is NOT a stop word and NOT a punctuation mark
filtered_words = [token.text for token in doc if not token.is_stop and not token.is_punct]

# --- Output Section ---
print("Original Text:\n", text)
print("\nFiltered Text (No Stop Words):\n", " ".join(filtered_words))

print("\nIdentified Stop Words in the text:")
for token in doc:
    if token.is_stop:
        print(f"- {token.text}")
```

```python
import re

def extract_register_numbers(text):
    pattern = r'\b(\d{2,4})([A-Za-z]{2,5})(\d{2,4})\b'
    matches = re.finditer(pattern, text)
    results = []

    for match in matches:
        results.append({
            "Register_Number": match.group(0),
            "Year": match.group(1),
            "Department": match.group(2).upper(),
            "Roll_No": match.group(3)
        })
    return results

# 1. Prompts you for input in Colab
paragraph = input("Paste or type your paragraph here: ")
paragraph2 = input("Paste or type your paragraph here: ")

# Combine both paragraphs
combined_paragraph = f"{paragraph}\n{paragraph2}"

# 2. Processes the input text
found_ids = extract_register_numbers(combined_paragraph)

if found_ids:
    print(f"\nFound {len(found_ids)} Register Number(s):\n")
    for item in found_ids:
        print(f"• ID: {item['Register_Number']} | Batch/Year: {item['Year']} | Dept: {item['Department']} | Roll No: {item['Roll_No']}")
else:
    print("\nNo register number patterns found in the text.")
```

```python
def make_ngrams(tokens, n):
    return [" ".join(tokens[i : i + n]) for i in range(len(tokens) - n + 1)]

sentence = input("Enter a sentence: ").split()
print("unigrams:", make_ngrams(sentence, 1))
print("bigrams:", make_ngrams(sentence, 2))
print("trigrams:", make_ngrams(sentence, 5))
```

```python
import re

POSITIVE = {
    "meeting", "schedule", "project", "attached", "report",
    "invoice", "thanks", "regards", "discussion", "agenda"
}

NEGATIVE = {
    "win", "winner", "prize", "free", "urgent",
    "guaranteed", "claim", "offer", "cash", "click"
}

def rule_based_sentiment(text):
    words = re.findall(r"\b[a-z]+\b|\$\d+|https?://\S+|!{2,}|[A-Z]{4,}", text)
    score = sum(w.lower() in POSITIVE or w.isupper() or "$" in w or "!" in w for w in words) - sum(w.lower() in NEGATIVE for w in words)
    return "positive" if score > 0 else "negative" if score < 0 else "neutral"


tests = [
    "URGENT! You have won a FREE $1000 cash prize! Click here to claim now!!!",
    "Hi team, please find attached the meeting schedule and project update report for tomorrow.",
    "Special discount offer for you, check out our website at https://example.com",
    "Thanks for your reply. Regards, John."
]

for t in tests:
    print(f"{rule_based_sentiment(t):9} | {t}")
```

```python
import speech_recognition as sr
import spacy
import os
import subprocess

# 1. Load the spaCy NLP model
nlp = spacy.load("en_core_web_sm")

# Original recorded path
original_audio_path = "/content/sample_data/recorded_audio.wav"
# Converted PCM WAV path
converted_audio_path = "/content/sample_data/recorded_audio_converted.wav"

try:
    if not os.path.exists(original_audio_path):
        raise FileNotFoundError

    print("🔄 Converting browser recording to standard PCM WAV...")
    # Use ffmpeg (pre-installed in Colab) to convert the container to true PCM WAV
    subprocess.run([
        'ffmpeg', '-y', '-i', original_audio_path,
        '-acodec', 'pcm_s16le', '-ar', '16000', '-ac', '1',
        converted_audio_path
    ], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

    recognizer = sr.Recognizer()

    with sr.AudioFile(converted_audio_path) as source:
        print("📁 Loading your converted voice...")
        audio_data = recognizer.record(source)

        print("🔄 Converting speech to text...")
        text = recognizer.recognize_google(audio_data)
        print(f"\n[Transcribed Text]: {text}")

        # --- 3. Pass the Transcribed Text to the NLP Pipeline ---
        doc = nlp(text)

        print("\n--- NLP Analysis ---")
        print("Tokens and Parts of Speech:")
        for token in doc:
            print(f"  {token.text} -> {token.pos_}")

        if doc.ents:
            print("\nNamed Entities Found:")
            for ent in doc.ents:
                print(f"  {ent.text} -> {ent.label_}")
        else:
            print("\nNamed Entities Found: None")

except FileNotFoundError:
    print(f"❌ Error: The file was not found at '{original_audio_path}'. Please record your voice using the recorder first!")
except sr.UnknownValueError:
    print("❌ Speech Recognition could not understand the audio. Try speaking louder or clearer.")
except sr.RequestError as e:
    print(f"❌ Could not connect to Google Speech Recognition service; {e}")
except Exception as e:
    print(f"❌ An unexpected error occurred: {e}")
```
